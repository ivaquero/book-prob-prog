### The logistic model applied to the iris dataset

```python
df = iris.query("species == ('setosa', 'versicolor')")
y_0 = pd.Categorical(df["species"]).codes
x_0 = df["sepal_length"].values
x_c = x_0 - x_0.mean()
```

```python
path = "../data/bap-ch04-iris-logreg.nc"

with pm.Model() as model_0:
    α = pm.Normal('α', mu=0, sigma=10)
    β = pm.Normal('β', mu=0, sigma=10)
    
    μ = α + pm.math.dot(x_c, β)    
    θ = pm.Deterministic('θ', pm.math.sigmoid(μ))
    bd = pm.Deterministic('bd', -α/β)
    
    yl = pm.Bernoulli('yl', p=θ, observed=y_0)

    if os.path.isfile(path):
        trace_0 = az.from_netcdf(path)
    else:
        trace_0 = pm.sample(1000)
        trace_0.to_netcdf(path)
```

```python
sun_0 = az.summary(trace_0, ['α', 'β', 'bd'])
sun_0.to_csv('../data/bap-ch04-iris-logreg.csv')
sun_0
```

```python
θs = trace_0.posterior["θ"]
θ = θs.mean(axis=0).mean(axis=0)
bds = trace_0.posterior["bd"]
idx = np.argsort(x_c)
bd_hdi = az.hdi(bds).bd

_, ax = plt.subplots()

ax.scatter(
    x_c,
    np.random.normal(y_0, 0.02),
    marker=".",
    color=[f"C{x}" for x in y_0],
)

ax.vlines(bds.mean(), 0, 1, color="k")
ax.fill_betweenx([0, 1], bd_hdi[0], bd_hdi[1], color="k", alpha=0.5)
ax.plot(x_c[idx], θ[idx], color="C2", lw=1)
az.plot_hdi(x_c, θs, color="C2")

ax.set(
    xlabel="sepal_length",
    ylabel="θ",
    xticks=ax.get_xticks(),
    xticklabels=np.arange(4, 8, 0.5),
);

# plt.savefig("../images/datasets/iris-logreg-hdi.png")
```

# Robust logistic regression

```python
df = iris.query("species == ('setosa', 'versicolor')") 
y_0 = pd.Categorical(df['species']).codes 
x_0 = df['sepal_length'].values 
y_0 = np.concatenate((y_0, np.ones(6, dtype=int))) 
x_0 = np.concatenate((x_0, [4.2, 4.5, 4.0, 4.3, 4.2, 4.4])) 
x_c = x_0 - x_0.mean()

_, ax = plt.subplots()
ax.plot(x_c, y_0, 'o', color='k');
```

```python
path = "../data/bap-ch04-iris-logreg-robust.nc"

with pm.Model() as model_rlg:
    α = pm.Normal('α', mu=0, sigma=10)
    β = pm.Normal('β', mu=0, sigma=10)
    
    μ = α + x_c *  β  
    θ = pm.Deterministic('θ', pm.math.sigmoid(μ))
    bd = pm.Deterministic('bd', -α/β)
    
    π = pm.Beta('π', 1., 1.) 
    p = π * 0.5 + (1 - π) * θ 
    
    yl = pm.Bernoulli('yl', p=p, observed=y_0)
    if os.path.isfile(path):
        trace_rlg = az.from_netcdf(path)
    else:
        trace_rlg = pm.sample(1000)
        trace_rlg.to_netcdf(path)
```

```python
# az.plot_trace(trace_rlg, varnames, compact=False);
```

```python
sum_rlg = az.summary(trace_rlg, ['α', 'β', 'bd'])
sum_rlg.to_csv("../data/bap-ch04-iris-logreg-robust.csv")
sum_rlg
```

```python
θs = trace_rlg.posterior['θ']
θ = θs.mean(axis=0).mean(axis=0)
bds = trace_rlg.posterior['bd']
bd = bds.mean()
bd_hdi = az.hdi(bds).bd
idx = np.argsort(x_c)

_, ax = plt.subplots()

ax.scatter(x_c, np.random.normal(y_0, 0.02), marker='.', color=[f'C{x}' for x in y_0])

ax.plot(x_c[idx], θ[idx], color='C2', lw=3);
ax.vlines(bd, 0, 1, color='k')
ax.fill_betweenx([0, 1], bd_hdi[0], bd_hdi[1], color='k', alpha=0.5)

θ_hdi = az.hdi(θs).θ[idx]
ax.fill_between(x_c[idx], θ_hdi[:,0], θ_hdi[:,1], color='C2', alpha=0.5)

ax.set(
    xlabel="sepal_length",
    ylabel="θ",
    xticks=ax.get_xticks(),
    xticklabels=np.arange(3.5, 8, 0.5),
);
# plt.savefig("../images/datasets/iris-logreg-robust-hdi.png")
```

# Multiple logistic regression

```python
df = iris.query("species == ('setosa', 'versicolor')") 
y_1 = pd.Categorical(df['species']).codes
x_1 = df[['sepal_length', 'sepal_width'] ].values
```

```python
path = "../data/bap-ch04-iris-logreg-multi.nc"

with pm.Model() as model_1: 
    α = pm.Normal('α', mu=0, sigma=10) 
    β = pm.Normal('β', mu=0, sigma=2, shape=2) 
     
    μ = α + pm.math.dot(x_1, β) 
    θ = pm.Deterministic('θ', 1 / (1 + pm.math.exp(-μ))) 
    bd = pm.Deterministic('bd', -α/β[1] - β[0]/β[1] * x_1[:,0])
     
    yl = pm.Bernoulli('yl', p=θ, observed=y_1) 
 
    if os.path.isfile(path):
        trace_1 = az.from_netcdf(path)
    else:
        trace_1 = pm.sample(2000)
        trace_1.to_netcdf(path)
```

```python
az.plot_forest(trace_1, var_names=['α', 'β']);
```

```python
idx = np.argsort(x_1[:,0]) 
bds = trace_1.posterior['bd'].values
bd = bds.mean(0).mean(0)[idx]

_, ax = plt.subplots()

ax.scatter(x_1[:,0], x_1[:,1], c=[f'C{x}' for x in y_0]) 
ax.plot(x_1[:,0][idx], bd, color='k'); 
az.plot_hdi(x_1[:,0], bds, color='k')
 
ax.set(xlabel='sepal_length', ylabel='sepal_width')

# plt.savefig("../images/datasets/iris-logreg-multi.png")
```

## Interpreting the coefficients of a logistic regression

```python
prob = np.linspace(0.01, 1, 100)
odds = prob / (1 - prob)

_, ax = plt.subplots()

ax2 = ax.twinx()
ax.plot(prob, odds, 'C0')
ax2.plot(prob, np.log(odds), 'C1')

ax.set_xlabel('probability')
ax.set_ylabel('odds', color='C0')
ax2.set_ylabel('logit', color='C1')

# plt.savefig("../images/funcs/logit-odds.png")
```

```python
sum_1 = az.summary(trace_1, var_names=['α', 'β'])
sum_1.to_csv("../data/bap-ch04-iris-logreg-multi.csv")
sum_1
```

```python
x_1 = 4.5  # sepal_length
x_2 = 3   # sepal_width 

log_odds_versicolor_i = (sum_1['mean'] * [1, x_1, x_2]).sum()
probability_versicolor_i = logistic(log_odds_versicolor_i)


log_odds_versicolor_f = (sum_1['mean'] * [1, x_1 + 1, x_2]).sum()
probability_versicolor_f = logistic(log_odds_versicolor_f)

log_odds_versicolor_f - log_odds_versicolor_i, probability_versicolor_f - probability_versicolor_i
```

## Dealing with unbalanced classes

```python
df = iris.query("species == ('setosa', 'versicolor')")[45:]
y_3 = pd.Categorical(df['species']).codes
x_3 = df[['sepal_length', 'sepal_width']].values
```

```python
path = "../data/bap-ch04-iris-unbalanced.nc"

with pm.Model() as model_3: 
    α = pm.Normal('α', mu=0, sigma=10) 
    β = pm.Normal('β', mu=0, sigma=2, shape=2) 
     
    μ = α + pm.math.dot(x_3, β) 
    θ = 1 / (1 + pm.math.exp(-μ)) 
    bd = pm.Deterministic('bd', -α/β[1] - β[0]/β[1] * x_3[:,0]) 
    yl = pm.Bernoulli('yl', p=θ, observed=y_3) 
 
    if os.path.isfile(path):
        trace_3 = az.from_netcdf(path)
    else:
        trace_3 = pm.sample(1000)
        trace_3.to_netcdf(path)
```

```python
# az.plot_trace(trace_3, varnames, compact=False);
```

```python
idx = np.argsort(x_3[:,0]) 
bds = trace_3.posterior['bd'].values
bd = bds.mean(0).mean(0)[idx]

_, ax = plt.subplots()

ax.scatter(x_3[:,0], x_3[:,1], c= [f'C{x}' for x in y_3]) 
ax.plot(x_3[:,0][idx], bd, color='k')

az.plot_hdi(x_3[:,0], bds, color='k')
ax.set(xlabel='sepal_length', ylabel='sepal_width')
# plt.savefig("../images/datasets/iris-unbalanced.png")
```

# Softmax regression

```python
y_s = pd.Categorical(iris['species']).codes
x_s = iris[iris.columns[:-1]].values
x_s = (x_s - x_s.mean(axis=0)) / x_s.std(axis=0)
```

```python
path = "../data/bap-ch04-iris-softmax.nc"

with pm.Model() as model_s:
    α = pm.Normal('α', mu=0, sigma=5, shape=3)
    β = pm.Normal('β', mu=0, sigma=5, shape=(4,3))
    μ = pm.Deterministic('μ', α + pm.math.dot(x_s, β))
    θ = pm.math.softmax(μ, axis=1)
    
    yl = pm.Categorical('yl', p=θ, observed=y_s)
    
    if os.path.isfile(path):
        trace_s = az.from_netcdf(path)
    else:
        trace_s = pm.sample(2000)
        trace_s.to_netcdf(path)
```

```python
# az.plot_forest(trace_s, var_names=['α', 'β']);
```

```python
# az.summary(trace_s, var_names=['α', 'β'])
```

```python
data_pred = trace_s.posterior['μ'].values.mean(0).mean(0)
y_pred = [np.exp(point)/np.sum(np.exp(point), axis=0) for point in data_pred]

f'{np.sum(y_s == np.argmax(y_pred, axis=1)) / len(y_s):.2f}'
```

```python
path = "../data/bap-ch04-iris-softmax-f.nc"

with pm.Model() as model_sf:
    α = pm.Normal('α', mu=0, sigma=2, shape=2)
    β = pm.Normal('β', mu=0, sigma=2, shape=(4, 2))
    α_f = at.concatenate([[0] ,α])
    β_f = at.concatenate([np.zeros((4,1)) , β], axis=1)
    μ = α_f + pm.math.dot(x_s, β_f)
    θ = pm.math.softmax(μ, axis=1)
    
    yl = pm.Categorical('yl', p=θ, observed=y_s)
    if os.path.isfile(path):
        trace_sf = az.from_netcdf(path)
    else:
        trace_sf = pm.sample(1000)
        trace_sf.to_netcdf(path)
```

```python
# az.plot_forest(trace_sf, var_names=['α', 'β']);
```

```python
az.summary(trace_sf, var_names=['α', 'β'])
```

# Poisson regression

## The Zero-Inflated Poisson model

```python
n = 100
θ_real = 2.5
ψ = 0.1

# Simulate some data
counts = np.array(
    [
        (np.random.random() > (1 - ψ)) * np.random.poisson(θ_real)
        for _ in range(n)
    ]
)
```

```python
path = "../data/bap-ch04-zip.nc"

with pm.Model() as ZIP:
    ψ = pm.Beta('ψ', 1, 1)
    θ = pm.Gamma('θ', 2, 0.1)
    y = pm.ZeroInflatedPoisson('y', ψ, θ, observed=counts)
    
    if os.path.isfile(path):
        trace_z = az.from_netcdf(path)
    else:
        trace_z = pm.sample(1000)
        trace_z.to_netcdf(path)
```

## Poisson regression and ZIP regression

```python
fish = pd.read_csv('../data/fish.csv')
fish.head()
```

```python
path = "../data/bap-ch04-zipreg.nc"

with pm.Model() as ZIP_reg:
    ψ = pm.Beta('ψ', 1, 1)
    α = pm.Normal('α', 0, 10)
    β = pm.Normal('β', 0, 10, shape=2)
    θ = pm.math.exp(α + β[0] * fish['child'] + β[1] * fish['camper'])
    
    yl = pm.ZeroInflatedPoisson('yl', ψ, θ, observed=fish['count'])

    if os.path.isfile(path):
        trace_zreg = az.from_netcdf(path)
    else:
        trace_zreg = pm.sample(1000)
        trace_zreg.to_netcdf(path)
```
