## Simple linear regression

```python
np.random.seed(1)
N = 100
α_real = 2.5
β_real = 0.9
ϵ_real = np.random.normal(0, 0.5, size=N)

x = np.random.normal(10, 1, N)
y_real = α_real + β_real * x
y = y_real + ϵ_real

# we can center the data
#x = x - x.mean()
# or standardize the data
#x = (x - x.mean())/x.std()
#y = (y - y.mean())/y.std()
```

```python
_, ax = plt.subplots(1, 2, figsize=(8, 4), constrained_layout=True)

ax[0].plot(x, y, '.')
ax[0].plot(x, y_real, 'k')
ax[0].set(xlabel='x', ylabel='y')

az.plot_kde(y, ax=ax[1])
ax[1].set_xlabel('y')

# plt.savefig("../images/books/bap-ch03-linreg-simple.png")
```

```python
path = "../data/bap-ch03-linreg.nc"

with pm.Model() as model_linreg:
    α = pm.Normal('α', mu=0, sigma=10)
    β = pm.Normal('β', mu=0, sigma=1)
    ϵ = pm.HalfCauchy('ϵ', 5)

    μ = pm.Deterministic('μ', α + β * x)
    y_pred = pm.Normal('y_pred', mu=μ, sigma=ϵ, observed=y)

    if os.path.isfile(path):
        trace_linreg = az.from_netcdf(path)
    else:
        trace_linreg = pm.sample(2000, tune=1000)
        trace_linreg.to_netcdf(path)
```

```python
az.plot_trace(trace_linreg, var_names=['α', 'β', 'ϵ'], compact=False);
# plt.savefig("../images/books/bap-ch03-linreg-trace.png")
```

### Modyfing the data before running the models

```python
az.plot_pair(trace_linreg, var_names=['α', 'β'], scatter_kwargs={'alpha': 0.1})
# plt.savefig("../images/books/bap-ch03-linreg-autocorr.png")
```

### Interpreting the posterior

```python
αs = trace_linreg.posterior["α"].values
βs = trace_linreg.posterior["β"].values

α_m = αs.mean()
β_m = βs.mean()
```

```python
_, ax = plt.subplots()

ax.plot(x, y, "C0.")

draws = range(0, len(αs), 10)
ax.plot(
    x,
    αs[draws] + βs[draws] * x[:, np.newaxis],
    c="gray",
    alpha=0.5,
)

ax.plot(
    x,
    α_m + β_m * x,
    c="k",
    label=f"y = {α_m:.2f} + {β_m:.2f} * x",
)

ax.set(xlabel="x", ylabel="y")
ax.legend()
```

```python
_, ax = plt.subplots()

ax.plot(x, y, "b.")
ax.plot(
    x,
    α_m + β_m * x,
    c="k",
    label=f"y = {α_m:.2f} + {β_m:.2f} * x",
)

sig = az.plot_hdi(
    x,
    trace_linreg.posterior["μ"],
    color="gray",
    ax=ax
)

ax.set(xlabel="x", ylabel="y")
ax.legend()
# plt.savefig("../images/books/bap-ch03-linreg-hdi.png")
```

```python
ppc = pm.sample_posterior_predictive(
    trace_linreg.sel(draw=slice(None, 1999, 1)),
    model=model_linreg,
)
```

```python
_, ax = plt.subplots()

ax.plot(x, y, "b.")
ax.plot(
    x,
    α_m + β_m * x,
    c="k",
    label=f"y = {α_m:.2f} + {β_m:.2f} * x",
)

az.plot_hdi(
    x,
    ppc.posterior_predictive["y_pred"],
    hdi_prob=0.5,
    color="gray",
    ax=ax,
)
az.plot_hdi(
    x,
    ppc.posterior_predictive["y_pred"],
    color="gray",
    ax=ax,
)

ax.set(xlabel="x", ylabel="y")
# plt.savefig("../images/books/bap-ch03-linreg-hdi-ppc.png")
```

## Robust linear regression

```python
ans = pd.read_csv('../data/anscombe.csv')

x_3 = ans[ans.group == 'III']['x'].values
y_3 = ans[ans.group == 'III']['y'].values
x_3 = x_3 - x_3.mean()
```

```python
_, ax = plt.subplots(1, 2, figsize=(8, 4))

β_c, α_c = stats.linregress(x_3, y_3)[:2]

ax[0].plot(x_3, (α_c + β_c * x_3), "k", label=f"y ={α_c:.2f} + {β_c:.2f} * x")
ax[0].plot(x_3, y_3, "C0o")
ax[0].set(xlabel="x", ylabel="y")
ax[0].legend(loc=0)

az.plot_kde(y_3, ax=ax[1], rug=True)
ax[1].set(xlabel="y", yticks=[])
# plt.savefig("../images/books/bap-ch03-linreg-anscombe.png")
```

```python
path = "../data/bap-ch03-anscombe.nc"

with pm.Model() as anscombe_t:
    α = pm.Normal('α', mu=y_3.mean(), sigma=1)
    β = pm.Normal('β', mu=0, sigma=1)
    ϵ = pm.HalfNormal('ϵ', 5)
    ν_ = pm.Exponential('ν_', 1/29)
    ν = pm.Deterministic('ν', ν_ + 1)

    y_pred = pm.StudentT('y_pred', mu=α + β * x_3,
                         sigma=ϵ, nu=ν, observed=y_3)

    if os.path.isfile(path):
        trace_t = az.from_netcdf(path)
    else:
        trace_t = pm.sample(2000)
        trace_t.to_netcdf(path)
```

```python
varnames = ['α', 'β', 'ϵ', 'ν']
az.plot_trace(trace_t, var_names=varnames, compact=False);
# plt.savefig("../images/books/bap-ch03-anscombe-trace.png")
```

```python
β_c, α_c = stats.linregress(x_3, y_3)[:2]

α_m = trace_t.posterior['α'].mean().values
β_m = trace_t.posterior['β'].mean().values

_, ax = plt.subplots(figsize=(5, 4))

ax.plot(x_3, y_3, 'C0o')
ax.plot(x_3, (α_c + β_c * x_3), c='k', label='non-robust', alpha=0.5)
ax.plot(x_3, (α_m + β_m * x_3), c='k', label='robust')
ax.set(xlabel="x", ylabel="y")
ax.legend()

plt.savefig("../images/books/bap-ch03-ans-robust.png")
```

```python
sum_t = az.summary(trace_t, var_names=varnames)
sum_t.to_csv("../data/bap-ch02-ans-trace.csv")
sum_t
```

```python
ppc = pm.sample_posterior_predictive(
    trace_t.sel(draw=slice(None, 199, 1)),
    model=anscombe_t,
    random_seed=2,
)
```

```python
ax = az.plot_ppc(ppc, mean=True)
ax.set(xlim=(2, 12))
ax.legend(loc='upper left', fontsize='medium')
# plt.savefig("../images/books/bap-ch03-ans-ppc.png")
```

### Multivariate Normal Distribution 

```python
import itertools

σ_x1 = 1
σs_x2 = [1, 2]
ρs = [-0.90, -0.5, 0, 0.5, 0.90]

k, l = np.mgrid[-5:5:0.1, -5:5:0.1]
pos = np.empty(k.shape + (2,))
pos[:, :, 0], pos[:, :, 1] = k, l

f, axes = plt.subplots(
    2, 5, sharex=True, sharey=True, figsize=(12, 6), constrained_layout=True
)

pairs = itertools.product(σs_x2, ρs)

for ax, pair in zip(axes.flatten(), pairs):
    σ_x2, ρ = pair
    cov = [
        [σ_x1**2, σ_x1 * σ_x2 * ρ],
        [σ_x1 * σ_x2 * ρ, σ_x2**2],
    ]
    ax.contour(k, l, stats.multivariate_normal([0, 0], cov).pdf(pos))
    ax.set(xlim=(-8, 8), ylim=(-8, 8), yticks=[-5, 0, 5])
    ax.plot(
        0,
        0,
        label=f"$σ_{{x2}}$ = {σ_x2:3.2f}\n$ρ$ = {ρ:3.2f}",
        alpha=0
    )
    ax.legend()

f.text(0.5, -0.05, "x_1", ha="center", fontsize="xx-large")
f.text(-0.02, 0.5, "x_2", va="center", fontsize="xx-large")

# plt.savefig("../images/books/bap-ch03-covmat.png")
```

```python
data = pd.read_csv('../data/babies.csv')
data.plot.scatter('Month', 'Lenght')
```

```python
with pm.Model() as model_vv:
    α = pm.Normal('α', sd=10)
    β = pm.Normal('β', sd=10)
    γ = pm.HalfNormal('γ', sd=10)
    δ = pm.HalfNormal('δ', sd=10)

    x_shared = shared(data.Month.values * 1.)

    μ = pm.Deterministic('μ', α + β * x_shared**0.5)
    ϵ = pm.Deterministic('ϵ', γ + δ * x_shared)

    y_pred = pm.Normal('y_pred', mu=μ, sd=ϵ, observed=data.Lenght)

    trace_vv = pm.sample(1000, tune=1000)
```

## Hierarchical linear regression

```python
N = 20
M = 8
idx = np.repeat(range(M - 1), N)
idx = np.append(idx, 7)
np.random.seed(314)

α_real = np.random.normal(2.5, 0.5, size=M)
β_real = np.random.beta(6, 1, size=M)
ϵ_real = np.random.normal(0, 0.5, size=len(idx))

y_m = np.zeros(len(idx))
x_m = np.random.normal(10, 1, len(idx))
y_m = α_real[idx] + β_real[idx] * x_m + ϵ_real

_, axes = plt.subplots(2, 4, sharex=True, sharey=True)

j, k = 0, N
axes = axes.flatten()
for i in range(M):
    axes[i].scatter(x_m[j:k], y_m[j:k])
    axes[i].set(
        xlabel=f"x_{i}",
        ylabel=f"y_{i}",
        xlim=(6, 15),
        ylim=(7, 17),
    )
    j += N
    k += N
```

```python
x_centered = x_m - x_m.mean()
```

```python
with pm.Model() as unpooled_model:
    α_tmp = pm.Normal("α_tmp", mu=0, sd=10, shape=M)
    β = pm.Normal("β", mu=0, sd=10, shape=M)
    ϵ = pm.HalfCauchy("ϵ", 5)
    ν = pm.Exponential("ν", 1 / 30)

    y_pred = pm.StudentT(
        "y_pred",
        mu=α_tmp[idx] + β[idx] * x_centered,
        sd=ϵ,
        nu=ν,
        observed=y_m,
    )

    α = pm.Deterministic("α", α_tmp - β * x_m.mean())

    trace_up = pm.sample(2000)
```

```python
az.plot_forest(trace_up, var_names=['α', 'β'], combined=True)
```

```python
with pm.Model() as hierarchical_model:
    # hyper-priors
    α_μ_tmp = pm.Normal('α_μ_tmp', mu=0, sd=10)
    α_σ_tmp = pm.HalfNormal('α_σ_tmp', 10)
    β_μ = pm.Normal('β_μ', mu=0, sd=10)
    β_σ = pm.HalfNormal('β_σ', sd=10)

    # priors
    α_tmp = pm.Normal('α_tmp', mu=α_μ_tmp, sd=α_σ_tmp, shape=M)
    β = pm.Normal('β', mu=β_μ, sd=β_σ, shape=M)
    ϵ = pm.HalfCauchy('ϵ', 5)
    ν = pm.Exponential('ν', 1/30)

    y_pred = pm.StudentT('y_pred', mu=α_tmp[idx] + β[idx] * x_centered,
                         sd=ϵ, nu=ν, observed=y_m)

    α = pm.Deterministic('α', α_tmp - β * x_m.mean())
    α_μ = pm.Deterministic('α_μ', α_μ_tmp - β_μ * x_m.mean())
    α_σ = pm.Deterministic('α_sd', α_σ_tmp - β_μ * x_m.mean())

    trace_hm = pm.sample(1000)
```
